```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import StandardScaler
```

```python
numerical_columns=['Age','BMI','Exercise_Hours_Per_Week']
categorical_column=['Gender','Smoking','Alcohol_Use','Family_History','Air_Pollution','Chronic_Cough','Chest_Pain','Fatigue','Cancer']
```

```python
#Train Test Split
x=df.iloc[:,0:11]
y=df.iloc[:,11]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.head()
```

```python
#Encoding the categorical column
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder



```

```python
x_train['Gender'].unique()
```

```python
labels=['Gender','Smoking','Alcohol_Use','Family_History','Chronic_Cough','Chest_Pain','Fatigue']
# Create a dictionary to store a LabelEncoder for each column
label_encoders = {}

for col in labels:
  le_instance = LabelEncoder()
  x_train[col] = le_instance.fit_transform(x_train[col])
  x_test[col] = le_instance.transform(x_test[col])
  label_encoders[col] = le_instance # Store the fitted encoder for later use
```

```python
lbe=LabelEncoder()
y_train=lbe.fit_transform(y_train)
y_test=lbe.fit_transform(y_test)
y_train=pd.DataFrame(y_train)
y_test=pd.DataFrame(y_test)
```

```python
y_train.head()
```

```python
x_train.head()
```

```python
oe=OrdinalEncoder(categories=[["Low","Medium","High"]])
oe.fit(x_train[["Air_Pollution"]])
x_train[["Air_Pollution"]]=oe.transform(x_train[["Air_Pollution"]])
x_test[["Air_Pollution"]]=oe.transform(x_test[["Air_Pollution"]])
```

```python
x_train.head()
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()

# Store original column names before transformation
original_x_train_cols = x_train.columns
original_x_test_cols = x_test.columns

# Transform and convert back to DataFrame with original column names
x_train_scaled = sc.fit_transform(x_train)
x_train = pd.DataFrame(x_train_scaled, columns=original_x_train_cols, index=x_train.index)

x_test_scaled = sc.transform(x_test) # Use transform, not fit_transform for test set
x_test = pd.DataFrame(x_test_scaled, columns=original_x_test_cols, index=x_test.index)
```

```python
from sklearn.neighbors import KNeighborsClassifier
best_k=0
best_accuracy=0
for k in range(1,21):
  knn=KNeighborsClassifier(n_neighbors=k,metric='manhattan',weights='distance',algorithm='brute')
  scores=cross_val_score(knn,x_train,y_train,cv=10,scoring='accuracy')
  accuracy=scores.mean()
  if accuracy>best_accuracy:
    best_accuracy=accuracy
    best_k=k

```

```python
def predict_patient(age, bmi, gender, smoking, alcohol,
                    family_history, air_pollution,
                    chronic_cough, chest_pain, fatigue,
                    exercise_hours):

    # 1. Create new patient data
    new_patient = pd.DataFrame([{
        'Age': age,
        'BMI': bmi,
        'Gender': gender,
        'Smoking': smoking,
        'Alcohol_Use': alcohol,
        'Family_History': family_history,
        'Air_Pollution': air_pollution,
        'Chronic_Cough': chronic_cough,
        'Chest_Pain': chest_pain,
        'Fatigue': fatigue,
        'Exercise_Hours_Per_Week': exercise_hours
    }])

    # 2. Encode categorical columns
    categorical_columns = [
        'Gender',
        'Smoking',
        'Alcohol_Use',
        'Family_History',
        'Chronic_Cough',
        'Chest_Pain',
        'Fatigue'
    ]

    for col in categorical_columns:
        new_patient[col] = label_encoders[col].transform(new_patient[col])

    # 3. Encode Air Pollution
    new_patient[['Air_Pollution']] = oe.transform(
        new_patient[['Air_Pollution']]
    )

    # 4. Scale the data
    new_patient_scaled = sc.transform(new_patient)

    # 5. Predict
    prediction = model.predict(new_patient_scaled)

    # 6. Convert prediction back to Yes/No
    result = lbe.inverse_transform(prediction)

    return result[0]
```

```python
result = predict_patient(
    45,          # Age
    28.1,        # BMI
    'Female',    # Gender
    'No',        # Smoking
    'Yes',       # Alcohol Use
    'No',        # Family History
    'High',      # Air Pollution
    'Yes',       # Chronic Cough
    'No',        # Chest Pain
    'Yes',       # Fatigue
    1.5          # Exercise Hours
)

print("Cancer Prediction:", result)
```
